# 1. Import Library

Pada tahap ini dilakukan import library yang digunakan untuk pengolahan data dan visualisasi.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

# 2. Membuat Pandas Series

Series adalah struktur data satu dimensi pada Pandas yang mirip seperti array, tetapi memiliki index.

```python
nilai = pd.Series([80, 75, 90, 85, 88])

print("Data Series Nilai:")
print(nilai)

print("\nRata-rata nilai:", nilai.mean())
print("Nilai maksimum:", nilai.max())
print("Nilai minimum:", nilai.min())
```

# 3. Membuat DataFrame

DataFrame adalah struktur data dua dimensi yang terdiri dari baris dan kolom. DataFrame sering digunakan untuk menyimpan dan mengolah dataset.

```python
data_mahasiswa = {
    "Nama": ["Amin", "Budi", "Citra", "Dewi", "Eka", "Fajar"],
    "NIM": ["220401010124", "220401010125", "220401010126", "220401010127", "220401010128", "220401010129"],
    "Kelas": ["IF405", "IF405", "IF403", "IF403", "IF401", "IF401"],
    "Nilai_Tugas": [85, 78, 90, 88, 75, 82],
    "Nilai_UTS": [88, 80, 92, 84, 78, 85],
    "Kehadiran": [14, 13, 15, 14, 12, 13]
}

df = pd.DataFrame(data_mahasiswa)
df
```

# 4. Melihat Informasi DataFrame

Tahap ini digunakan untuk melihat struktur data, tipe data, ukuran dataset, dan beberapa baris awal.

```python
print("Ukuran DataFrame:", df.shape)

print("\nInformasi DataFrame:")
df.info()

print("\nData awal:")
display(df.head())
```

# 5. Statistik Deskriptif

Statistik deskriptif digunakan untuk melihat ringkasan nilai numerik seperti rata-rata, standar deviasi, nilai minimum, dan maksimum.

```python
df.describe().round(2)
```

# 6. Seleksi Kolom

Pada bagian ini dilakukan pemilihan satu kolom atau beberapa kolom dari DataFrame.

```python
print("Kolom Nama:")
display(df["Nama"])

print("\nKolom Nama dan Nilai UTS:")
display(df[["Nama", "Nilai_UTS"]])
```

# 7. Seleksi Baris dengan loc dan iloc

`loc` digunakan untuk memilih data berdasarkan label/index, sedangkan `iloc` digunakan untuk memilih data berdasarkan posisi baris dan kolom.

```python
print("Menggunakan loc:")
display(df.loc[0:2, ["Nama", "Kelas", "Nilai_UTS"]])

print("\nMenggunakan iloc:")
display(df.iloc[0:3, 0:4])
```

# 8. Filtering Data

Filtering digunakan untuk mengambil data berdasarkan kondisi tertentu. Contohnya menampilkan mahasiswa dengan nilai UTS lebih dari atau sama dengan 85.

```python
filter_uts = df[df["Nilai_UTS"] >= 85]

print("Mahasiswa dengan Nilai UTS >= 85:")
display(filter_uts)
```

# 9. Menambahkan Kolom Baru

Pada tahap ini ditambahkan kolom `Nilai_Akhir` dan `Status` berdasarkan hasil perhitungan dari nilai tugas dan nilai UTS.

```python
df["Nilai_Akhir"] = (df["Nilai_Tugas"] * 0.4) + (df["Nilai_UTS"] * 0.6)

df["Status"] = np.where(df["Nilai_Akhir"] >= 80, "Lulus", "Perlu Perbaikan")

df
```

# 10. Mengurutkan Data

Sorting digunakan untuk mengurutkan data berdasarkan kolom tertentu. Pada contoh ini data diurutkan berdasarkan `Nilai_Akhir` dari yang tertinggi.

```python
df_sorted = df.sort_values(by="Nilai_Akhir", ascending=False)

df_sorted
```

# 11. Grouping Data Berdasarkan Kelas

Grouping digunakan untuk mengelompokkan data berdasarkan kategori tertentu. Pada contoh ini data dikelompokkan berdasarkan kelas.

```python
group_kelas = df.groupby("Kelas").agg({
    "Nilai_Tugas": "mean",
    "Nilai_UTS": "mean",
    "Nilai_Akhir": "mean",
    "Kehadiran": "mean"
}).round(2)

group_kelas
```

# 12. Mengecek Missing Value

Missing value perlu diperiksa agar data yang digunakan dalam analisis tidak memiliki nilai kosong.

```python
print("Jumlah missing value setiap kolom:")
print(df.isnull().sum())
```

# 13. Menyimpan DataFrame ke File CSV

DataFrame dapat disimpan dalam bentuk file CSV agar dapat digunakan kembali pada proses analisis berikutnya.

```python
output_file = "data_mahasiswa_pertemuan2.csv"
df.to_csv(output_file, index=False)

print(f"Data berhasil disimpan ke file: {output_file}")
```

# 14. Membaca Kembali File CSV

Setelah data disimpan, file CSV dapat dibaca kembali menggunakan fungsi `read_csv()`.

```python
df_read = pd.read_csv("data_mahasiswa_pertemuan2.csv")

df_read
```

# 15. Visualisasi Nilai Akhir Mahasiswa

Visualisasi barplot digunakan untuk membandingkan nilai akhir setiap mahasiswa.

```python
plt.figure(figsize=(9, 4))
sns.barplot(data=df, x="Nama", y="Nilai_Akhir")
plt.title("Nilai Akhir Mahasiswa")
plt.xlabel("Nama Mahasiswa")
plt.ylabel("Nilai Akhir")
plt.ylim(0, 100)
plt.show()
```

# 16. Visualisasi Rata-rata Nilai Berdasarkan Kelas

Visualisasi ini digunakan untuk melihat perbandingan rata-rata nilai akhir pada setiap kelas.

```python
group_kelas_reset = group_kelas.reset_index()

plt.figure(figsize=(7, 4))
sns.barplot(data=group_kelas_reset, x="Kelas", y="Nilai_Akhir")
plt.title("Rata-rata Nilai Akhir Berdasarkan Kelas")
plt.xlabel("Kelas")
plt.ylabel("Rata-rata Nilai Akhir")
plt.ylim(0, 100)
plt.show()
```

# 17. Analisis Korelasi Data Numerik

Korelasi digunakan untuk melihat hubungan antar variabel numerik dalam dataset.

```python
numeric_cols = df.select_dtypes(include="number").columns

corr = df[numeric_cols].corr()

plt.figure(figsize=(8, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1)
plt.title("Heatmap Korelasi Data Numerik")
plt.show()

display(corr.round(3))
```

# 1. Import Library

Pada tahap ini dilakukan import library yang dibutuhkan untuk pengolahan data, analisis data, dan visualisasi.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

# 2. Membuat Dataset

Dataset dibuat secara sederhana untuk simulasi proses data cleaning. Dataset berisi data pelanggan dengan beberapa masalah seperti missing value, duplikasi, format teks tidak konsisten, dan outlier.

```python
data = {
    "CustomerID": [1, 2, 3, 4, 5, 6, 7, 7, 8, 9, 10],
    "Nama": ["Ali", "Budi", "Citra", "Dewi", "Eka", "Fajar", "Gina", "Gina", "Hadi", "Indah", "Joko"],
    "Usia": [22, 35, np.nan, 41, 29, 33, 120, 120, 27, np.nan, 45],
    "Kota": ["jakarta", "Bandung", "JAKARTA", "surabaya", "bandung", "Medan", "jakarta", "jakarta", "Surabaya", "MEDAN", "Bandung"],
    "Pengeluaran": [150000, 300000, 250000, np.nan, 200000, 180000, 9000000, 9000000, 220000, 270000, np.nan],
    "Kategori": ["A", "B", "A", "C", "B", "A", "C", "C", "B", "A", "C"]
}

df = pd.DataFrame(data)
df
```

# 3. Melihat Informasi Dataset

Tahap ini digunakan untuk mengetahui jumlah baris, jumlah kolom, tipe data, dan tampilan awal dataset.

```python
print("Ukuran dataset:", df.shape)

print("\nInformasi dataset:")
df.info()

print("\nData awal:")
display(df.head())
```

# 4. Mengecek Missing Value

Missing value adalah data kosong atau data yang tidak memiliki nilai. Nilai kosong perlu ditangani agar tidak mengganggu proses analisis.

```python
missing_table = pd.DataFrame({
    "Jumlah Missing Value": df.isnull().sum(),
    "Persentase (%)": (df.isnull().mean() * 100).round(2)
})

missing_table
```

# 5. Menangani Missing Value

Missing value pada kolom numerik ditangani menggunakan median. Median digunakan karena lebih aman ketika terdapat nilai ekstrem atau outlier.

```python
df_clean = df.copy()

df_clean["Usia"] = df_clean["Usia"].fillna(df_clean["Usia"].median())
df_clean["Pengeluaran"] = df_clean["Pengeluaran"].fillna(df_clean["Pengeluaran"].median())

print("Missing value setelah dibersihkan:")
print(df_clean.isnull().sum())

df_clean
```

# 6. Mengecek dan Menghapus Data Duplikat

Data duplikat perlu dihapus agar tidak terjadi pengulangan data yang dapat memengaruhi hasil analisis.

```python
print("Jumlah data duplikat sebelum dibersihkan:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Jumlah data duplikat setelah dibersihkan:", df_clean.duplicated().sum())
print("Ukuran dataset setelah menghapus duplikat:", df_clean.shape)

df_clean
```

# 7. Normalisasi Teks

Normalisasi teks dilakukan untuk menyeragamkan format penulisan data kategorik. Contohnya, nama kota dibuat menjadi huruf kapital di awal kata.

```python
df_clean["Kota"] = df_clean["Kota"].str.strip().str.title()
df_clean["Kategori"] = df_clean["Kategori"].str.strip().str.upper()

df_clean[["CustomerID", "Nama", "Kota", "Kategori"]]
```

# 8. Statistik Deskriptif

Statistik deskriptif digunakan untuk melihat gambaran umum dari data numerik seperti rata-rata, nilai minimum, maksimum, dan standar deviasi.

```python
df_clean.describe().round(2)
```

# 9. Deteksi Outlier Menggunakan IQR

Outlier adalah nilai yang berbeda jauh dari data lainnya. Pada tahap ini digunakan metode Interquartile Range (IQR) untuk mendeteksi outlier.

```python
def detect_outlier_iqr(dataframe, column):
    q1 = dataframe[column].quantile(0.25)
    q3 = dataframe[column].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]

    return lower_bound, upper_bound, outliers

for col in ["Usia", "Pengeluaran"]:
    lower, upper, outliers = detect_outlier_iqr(df_clean, col)
    print(f"\nKolom: {col}")
    print(f"Batas bawah: {lower}")
    print(f"Batas atas : {upper}")
    print("Outlier:")
    display(outliers)
```

# 10. Menangani Outlier

Outlier dapat ditangani dengan beberapa cara. Pada praktikum ini, nilai outlier pada kolom `Usia` dan `Pengeluaran` diganti menggunakan batas atas IQR atau metode capping.

```python
df_final = df_clean.copy()

for col in ["Usia", "Pengeluaran"]:
    lower, upper, _ = detect_outlier_iqr(df_final, col)
    df_final[col] = np.where(df_final[col] > upper, upper, df_final[col])
    df_final[col] = np.where(df_final[col] < lower, lower, df_final[col])

df_final.describe().round(2)
```

# 11. Visualisasi Sebelum dan Sesudah Penanganan Outlier

Boxplot digunakan untuk membandingkan kondisi data sebelum dan sesudah outlier ditangani.

```python
for col in ["Usia", "Pengeluaran"]:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    sns.boxplot(y=df_clean[col], ax=axes[0])
    axes[0].set_title(f"Sebelum Capping - {col}")

    sns.boxplot(y=df_final[col], ax=axes[1])
    axes[1].set_title(f"Sesudah Capping - {col}")

    plt.tight_layout()
    plt.show()
```

# 12. Visualisasi Distribusi Data

Histogram digunakan untuk melihat sebaran data numerik setelah proses pembersihan data.

```python
for col in ["Usia", "Pengeluaran"]:
    plt.figure(figsize=(7, 4))
    sns.histplot(df_final[col], kde=True, bins=8)
    plt.title(f"Distribusi {col} Setelah Cleaning")
    plt.xlabel(col)
    plt.ylabel("Frekuensi")
    plt.show()
```

# 13. Analisis Data Kategorik

Countplot digunakan untuk melihat jumlah data berdasarkan kota dan kategori pelanggan.

```python
for col in ["Kota", "Kategori"]:
    plt.figure(figsize=(7, 4))
    sns.countplot(data=df_final, x=col)
    plt.title(f"Jumlah Data Berdasarkan {col}")
    plt.xlabel(col)
    plt.ylabel("Jumlah")
    plt.show()
```

# 14. Analisis Rata-rata Pengeluaran Berdasarkan Kota

Analisis ini digunakan untuk mengetahui rata-rata pengeluaran pelanggan pada setiap kota.

```python
avg_spending_city = df_final.groupby("Kota")["Pengeluaran"].mean().sort_values(ascending=False).reset_index()

display(avg_spending_city)

plt.figure(figsize=(8, 4))
sns.barplot(data=avg_spending_city, x="Kota", y="Pengeluaran")
plt.title("Rata-rata Pengeluaran Berdasarkan Kota")
plt.xlabel("Kota")
plt.ylabel("Rata-rata Pengeluaran")
plt.show()
```

# 15. Menyimpan Dataset Bersih

Dataset yang sudah dibersihkan dapat disimpan ke dalam file CSV agar dapat digunakan kembali pada analisis berikutnya.

```python
output_csv = "data_pelanggan_clean.csv"
df_final.to_csv(output_csv, index=False)

print(f"Dataset bersih berhasil disimpan sebagai: {output_csv}")

df_final
```

## List dan Perulangan

Pada bagian ini digunakan struktur data list dan perulangan `for`.

```python
# ================================================
# SEL 3: List dan perulangan
# ================================================

tools = ["Python", "Google Colab", "Jupyter Notebook", "GitHub"]

print("Tools Data Science yang akan dipelajari:")

for i, tool in enumerate(tools, 1):
    print(f"{i}. {tool}")
```

## Fungsi Sederhana

```python
# ================================================
# SEL 4: Fungsi sederhana
# ================================================

def perkenalan(nama, prodi, angkatan):
    return f"Hai! Saya {nama} dari {prodi}, angkatan {angkatan}."

hasil = perkenalan("Amin Siddik Rangkuti", "PJJ Informatika", 2022)

print(hasil)
```

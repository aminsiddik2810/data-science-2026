# 2. Variabel dan Tipe Data

Variabel digunakan untuk menyimpan nilai. Python memiliki beberapa tipe data dasar seperti string, integer, float, dan boolean.

```python
nama = "Amin Siddik Rangkuti"
nim = "220401010124"
umur = 28
ipk = 3.75
aktif = True

print("Nama:", nama)
print("NIM:", nim)
print("Umur:", umur)
print("IPK:", ipk)
print("Status Aktif:", aktif)

print("\nTipe data:")
print(type(nama))
print(type(nim))
print(type(umur))
print(type(ipk))
print(type(aktif))
```

# 3. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan operasi matematika seperti penjumlahan, pengurangan, perkalian, pembagian, pangkat, dan modulo.

```python
a = 20
b = 6

print("a =", a)
print("b =", b)

print("Penjumlahan:", a + b)
print("Pengurangan:", a - b)
print("Perkalian:", a * b)
print("Pembagian:", a / b)
print("Pembagian bulat:", a // b)
print("Modulo:", a % b)
print("Pangkat:", a ** b)
```

# 4. Struktur Data List

List digunakan untuk menyimpan banyak data dalam satu variabel. Data pada list dapat diakses menggunakan index.

```python
nilai = [80, 85, 90, 75, 88]

print("Data nilai:", nilai)
print("Nilai pertama:", nilai[0])
print("Nilai terakhir:", nilai[-1])
print("Jumlah data:", len(nilai))
print("Nilai tertinggi:", max(nilai))
print("Nilai terendah:", min(nilai))
print("Rata-rata:", sum(nilai) / len(nilai))
```

# 5. Struktur Data Dictionary

Dictionary digunakan untuk menyimpan data dalam bentuk pasangan key dan value.

```python
mahasiswa = {
    "nama": "Amin Siddik Rangkuti",
    "nim": "220401010124",
    "kelas": "IF405",
    "prodi": "PJJ Informatika"
}

print("Data mahasiswa:")
print(mahasiswa)

print("\nNama:", mahasiswa["nama"])
print("NIM:", mahasiswa["nim"])
print("Kelas:", mahasiswa["kelas"])
print("Program Studi:", mahasiswa["prodi"])
```

# 6. Percabangan If Else

Percabangan digunakan untuk menjalankan perintah tertentu berdasarkan kondisi yang diberikan.

```python
nilai_akhir = 85

if nilai_akhir >= 80:
    status = "Lulus dengan baik"
elif nilai_akhir >= 60:
    status = "Lulus"
else:
    status = "Tidak lulus"

print("Nilai akhir:", nilai_akhir)
print("Status:", status)
```

# 7. Perulangan For

Perulangan digunakan untuk menjalankan kode secara berulang. Pada contoh ini digunakan untuk menampilkan semua nilai dalam list.

```python
daftar_nilai = [80, 85, 90, 75, 88]

for i, nilai_mahasiswa in enumerate(daftar_nilai, start=1):
    print(f"Nilai ke-{i}: {nilai_mahasiswa}")
```

# 8. Fungsi Python

Fungsi digunakan untuk mengelompokkan kode agar dapat digunakan kembali. Pada contoh ini dibuat fungsi untuk menghitung rata-rata nilai.

```python
def hitung_rata_rata(data):
    return sum(data) / len(data)

rata_rata = hitung_rata_rata(daftar_nilai)

print("Daftar nilai:", daftar_nilai)
print("Rata-rata nilai:", rata_rata)
```

# 9. Import Library Data Science

Library seperti NumPy, Pandas, Matplotlib, dan Seaborn sering digunakan dalam Data Science untuk analisis data dan visualisasi.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("Library berhasil di-import")
```

# 10. Membuat Array dengan NumPy

NumPy digunakan untuk melakukan operasi numerik. Array NumPy memudahkan perhitungan matematika pada banyak data.

```python
data_numpy = np.array([70, 75, 80, 85, 90])

print("Array NumPy:", data_numpy)
print("Mean:", np.mean(data_numpy))
print("Median:", np.median(data_numpy))
print("Standar deviasi:", np.std(data_numpy))
print("Nilai maksimum:", np.max(data_numpy))
print("Nilai minimum:", np.min(data_numpy))
```

# 11. Membuat DataFrame dengan Pandas

DataFrame adalah struktur data berbentuk tabel yang terdiri dari baris dan kolom. DataFrame sangat penting dalam analisis data.

```python
data = {
    "Nama": ["Amin", "Budi", "Citra", "Dewi", "Eka"],
    "Nilai_Tugas": [85, 78, 90, 88, 75],
    "Nilai_UTS": [88, 80, 92, 84, 78],
    "Kehadiran": [14, 13, 15, 14, 12]
}

df = pd.DataFrame(data)

df
```

# 12. Informasi dan Statistik DataFrame

Tahap ini digunakan untuk melihat struktur data, tipe data, dan statistik deskriptif dari DataFrame.

```python
print("Informasi DataFrame:")
df.info()

print("\nStatistik deskriptif:")
display(df.describe().round(2))
```

# 13. Menambahkan Kolom Nilai Akhir

Kolom baru `Nilai_Akhir` dibuat dari kombinasi nilai tugas dan nilai UTS.

```python
df["Nilai_Akhir"] = (df["Nilai_Tugas"] * 0.4) + (df["Nilai_UTS"] * 0.6)

df
```

# 14. Menentukan Status Kelulusan

Status kelulusan ditentukan berdasarkan nilai akhir. Jika nilai akhir minimal 80, maka statusnya lulus.

```python
df["Status"] = np.where(df["Nilai_Akhir"] >= 80, "Lulus", "Perlu Perbaikan")

df
```

# 15. Visualisasi Nilai Akhir

Visualisasi digunakan agar data lebih mudah dipahami. Barplot berikut menampilkan nilai akhir setiap mahasiswa.

```python
plt.figure(figsize=(8, 4))
sns.barplot(data=df, x="Nama", y="Nilai_Akhir")
plt.title("Nilai Akhir Mahasiswa")
plt.xlabel("Nama Mahasiswa")
plt.ylabel("Nilai Akhir")
plt.ylim(0, 100)
plt.show()
```

# 16. Visualisasi Perbandingan Nilai Tugas dan UTS

Visualisasi ini digunakan untuk membandingkan nilai tugas dan nilai UTS setiap mahasiswa.

```python
df_melt = df.melt(
    id_vars="Nama",
    value_vars=["Nilai_Tugas", "Nilai_UTS"],
    var_name="Jenis_Nilai",
    value_name="Nilai"
)

plt.figure(figsize=(9, 4))
sns.barplot(data=df_melt, x="Nama", y="Nilai", hue="Jenis_Nilai")
plt.title("Perbandingan Nilai Tugas dan Nilai UTS")
plt.xlabel("Nama Mahasiswa")
plt.ylabel("Nilai")
plt.ylim(0, 100)
plt.show()
```

# 17. Analisis Sederhana

Pada tahap ini dilakukan analisis sederhana untuk mencari nilai akhir tertinggi, nilai akhir terendah, dan rata-rata kelas.

```python
nilai_tertinggi = df.loc[df["Nilai_Akhir"].idxmax()]
nilai_terendah = df.loc[df["Nilai_Akhir"].idxmin()]
rata_rata_kelas = df["Nilai_Akhir"].mean()

print("Mahasiswa dengan nilai akhir tertinggi:")
display(nilai_tertinggi)

print("\nMahasiswa dengan nilai akhir terendah:")
display(nilai_terendah)

print("\nRata-rata nilai akhir kelas:", round(rata_rata_kelas, 2))
```

# 18. Menyimpan Data ke CSV

DataFrame dapat disimpan ke file CSV agar dapat digunakan kembali pada proses analisis berikutnya.

```python
output_file = "data_pertemuan1.csv"
df.to_csv(output_file, index=False)

print(f"Data berhasil disimpan ke file: {output_file}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
```

```python
sns.set_theme(style='whitegrid', palette='Set2')

print("Library berhasil dimuat")
```

```python
df = sns.load_dataset('tips')

print("Dataset berhasil dimuat")
print(df.head())
```

```python
print("Jumlah baris dan kolom:")
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.describe().round(2))
```

```python
print(df.isnull().sum())
```

```python
fig = plt.figure(figsize=(16,12))

fig.suptitle(
    'Dashboard Analisis Dataset Tips',
    fontsize=18,
    fontweight='bold'
)

gs = gridspec.GridSpec(
    2,
    2,
    figure=fig,
    hspace=0.4,
    wspace=0.35
)

ax1=fig.add_subplot(gs[0,0])
ax2=fig.add_subplot(gs[0,1])
ax3=fig.add_subplot(gs[1,0])
ax4=fig.add_subplot(gs[1,1])
```

```python
avg_by_day=df.groupby('day')['total_bill'].mean()

avg_by_day.plot(
    kind='bar',
    ax=ax1,
    color='skyblue',
    edgecolor='black'
)

ax1.set_title(
    'Rata-rata Total Tagihan per Hari'
)

ax1.set_xlabel('Hari')
ax1.set_ylabel('Total Bill (USD)')

ax1.tick_params(
    axis='x',
    rotation=0
)

ax1.spines[['top','right']].set_visible(False)
```

```python
sns.histplot(
    data=df,
    x='total_bill',
    hue='time',
    kde=True,
    ax=ax2
)

ax2.set_title(
    'Distribusi Total Bill'
)

ax2.spines[['top','right']].set_visible(False)
```

```python
sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    hue='sex',
    ax=ax3
)

ax3.set_title(
    'Distribusi Tagihan per Hari dan Gender'
)

ax3.spines[['top','right']].set_visible(False)
```

```python
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='time',
    size='size',
    sizes=(30,180),
    alpha=0.7,
    ax=ax4
)

sns.regplot(
    data=df,
    x='total_bill',
    y='tip',
    scatter=False,
    color='red',
    ax=ax4
)

ax4.set_title(
    'Hubungan Total Bill dan Tip'
)

ax4.spines[['top','right']].set_visible(False)
```

```python
plt.savefig(
    'dashboard_tips.png',
    dpi=150,
    bbox_inches='tight'
)

plt.show()

print(
    "Dashboard berhasil disimpan: dashboard_tips.png"
)
```

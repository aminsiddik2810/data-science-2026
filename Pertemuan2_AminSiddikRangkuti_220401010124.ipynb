```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset Titanic berhasil dimuat.")
display(df.head())
```

```python
print("Informasi dataset:")
df.info()

print("\nStatistik deskriptif:")
display(df.describe())

print("\nJumlah missing values:")
print(df.isnull().sum())
```

```python
penumpang_selamat = df[df["Survived"] == 1]
penumpang_wanita = df[df["Sex"] == "female"]
wanita_kelas1 = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

print("Jumlah penumpang selamat:", len(penumpang_selamat))
print("Jumlah penumpang wanita:", len(penumpang_wanita))
print("Jumlah wanita kelas 1:", len(wanita_kelas1))

display(wanita_kelas1.head())
```

```python
contoh_loc = df.loc[df["Sex"] == "female", ["Name", "Age", "Sex", "Pclass", "Survived"]]

print("Contoh penggunaan .loc[]:")
display(contoh_loc.head())
```

```python
contoh_iloc = df.iloc[0:5, 0:5]

print("Contoh penggunaan .iloc[]:")
display(contoh_iloc)
```

```python
contoh_query = df.query("Sex == 'female' and Pclass == 1")

print("Contoh penggunaan query():")
display(contoh_query.head())
```

```python
print("Q1. Jumlah baris dan kolom dataset Titanic")

jumlah_baris, jumlah_kolom = df.shape
daftar_kolom = df.columns.tolist()

print("Jumlah baris:", jumlah_baris)
print("Jumlah kolom:", jumlah_kolom)
print("Daftar kolom:")
print(daftar_kolom)
```

```python
print("Q2. Missing values pada dataset Titanic")

missing_jumlah = df.isnull().sum()
missing_persen = (missing_jumlah / len(df)) * 100

missing_df = pd.DataFrame({
    "Jumlah Missing Values": missing_jumlah,
    "Persentase (%)": missing_persen.round(2)
})

missing_df = missing_df[missing_df["Jumlah Missing Values"] > 0]

display(missing_df)
```

```python
print("Q3. Persentase penumpang yang selamat")

persen_selamat = df["Survived"].mean() * 100

print(f"Persentase penumpang yang selamat: {persen_selamat:.2f}%")
```

```python
print("Q4. Penumpang wanita dari kelas 1")

wanita_kelas_1 = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]

jumlah_wanita_kelas_1 = len(wanita_kelas_1)
rata_usia_wanita_kelas_1 = wanita_kelas_1["Age"].mean()

print("Jumlah penumpang wanita kelas 1:", jumlah_wanita_kelas_1)
print(f"Rata-rata usia wanita kelas 1: {rata_usia_wanita_kelas_1:.2f} tahun")

display(wanita_kelas_1.head())
```

print("Q5. Tingkat keselamatan per kelas penumpang")

tingkat_selamat_per_kelas = df.groupby("Pclass")["Survived"].mean() * 100
tingkat_selamat_per_kelas = tingkat_selamat_per_kelas.round(2)

display(tingkat_selamat_per_kelas)

kelas_tertinggi = tingkat_selamat_per_kelas.idxmax()
nilai_tertinggi = tingkat_selamat_per_kelas.max()

print(f"Kelas dengan tingkat keselamatan tertinggi adalah Pclass {kelas_tertinggi} dengan persentase {nilai_tertinggi:.2f}%")

```python
print("Q6. Analisis tambahan berdasarkan jenis kelamin")

analisis_sex = df.groupby("Sex").agg(
    jumlah_penumpang=("PassengerId", "count"),
    jumlah_selamat=("Survived", "sum"),
    tingkat_selamat=("Survived", "mean"),
    rata_usia=("Age", "mean"),
    rata_tarif=("Fare", "mean")
)

analisis_sex["tingkat_selamat"] = (analisis_sex["tingkat_selamat"] * 100).round(2)
analisis_sex["rata_usia"] = analisis_sex["rata_usia"].round(2)
analisis_sex["rata_tarif"] = analisis_sex["rata_tarif"].round(2)

display(analisis_sex)

print("Interpretasi:")
print("Penumpang perempuan memiliki tingkat keselamatan lebih tinggi dibandingkan laki-laki.")
print("Variabel jenis kelamin menjadi salah satu faktor yang menarik dalam analisis keselamatan penumpang Titanic.")
```

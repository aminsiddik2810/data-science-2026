```python
missing_jumlah = df.isnull().sum()

missing_persen = (missing_jumlah / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    'Jumlah Missing Values': missing_jumlah,
    'Proporsi (%)': missing_persen
})

missing_summary
```

```python
persen_selamat = df['Survived'].mean() * 100

print(f"Persentase selamat: {persen_selamat:.2f}%")
```

```python
wanita_kelas1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]

print("Jumlah wanita kelas 1:", len(wanita_kelas1))

print("Rata-rata usia:", wanita_kelas1['Age'].mean())
```

```python
tingkat_selamat = df.groupby('Pclass')['Survived'].mean() * 100

print(tingkat_selamat)
```

```python
analisis_gender = df.groupby('Sex').agg(
    total_penumpang=('PassengerId', 'count'),
    jumlah_selamat=('Survived', 'sum'),
    tingkat_selamat=('Survived', 'mean')
)

analisis_gender
```

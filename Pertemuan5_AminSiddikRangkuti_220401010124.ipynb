# 1. Import Library

Pada tahap ini dilakukan import library yang dibutuhkan untuk pengolahan data, visualisasi, preprocessing, dan pembagian dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
```

# 2. Membuat Dataset

Dataset dibuat secara sederhana untuk simulasi proses preprocessing. Dataset berisi informasi pelanggan seperti umur, pendapatan, kategori produk, jumlah pembelian, dan status churn.

```python
data = {
    "CustomerID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Age": [23, 35, np.nan, 45, 31, 29, 40, np.nan, 50, 28],
    "Income": [3500000, 6000000, 4500000, np.nan, 5200000, 3900000, 7000000, 4800000, np.nan, 4100000],
    "Product_Category": ["Fashion", "Electronic", "Fashion", "Food", "Electronic", "Food", "Fashion", "Electronic", "Food", "Fashion"],
    "Purchase_Count": [2, 5, 3, 7, 4, 2, 8, 3, 6, 2],
    "Churn": ["No", "No", "Yes", "No", "Yes", "No", "No", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)
df
```

# 3. Melihat Informasi Dataset

Tahap ini bertujuan untuk memahami struktur dataset, tipe data, jumlah baris, jumlah kolom, dan ringkasan awal dari data.

```python
print("Ukuran dataset:", df.shape)

print("\nInformasi dataset:")
df.info()

print("\nLima data pertama:")
display(df.head())
```

# 4. Mengecek Missing Value

Missing value perlu dicek karena nilai kosong dapat memengaruhi hasil analisis dan model Machine Learning.

```python
missing_values = df.isnull().sum()
missing_percent = (df.isnull().mean() * 100).round(2)

missing_table = pd.DataFrame({
    "Jumlah Missing Value": missing_values,
    "Persentase (%)": missing_percent
})

missing_table
```

# 5. Menangani Missing Value

Missing value pada kolom numerik ditangani menggunakan nilai median. Median dipilih karena lebih tahan terhadap pengaruh outlier dibandingkan mean.

```python
df_clean = df.copy()

df_clean["Age"] = df_clean["Age"].fillna(df_clean["Age"].median())
df_clean["Income"] = df_clean["Income"].fillna(df_clean["Income"].median())

print("Missing value setelah imputasi:")
print(df_clean.isnull().sum())

df_clean
```

# 6. Mengecek Data Duplikat

Data duplikat perlu diperiksa agar tidak terjadi pengulangan data yang dapat memengaruhi hasil analisis.

```python
print("Jumlah data duplikat:", df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Ukuran dataset setelah menghapus duplikat:", df_clean.shape)
```

# 7. Statistik Deskriptif

Statistik deskriptif digunakan untuk mengetahui gambaran umum fitur numerik seperti nilai rata-rata, minimum, maksimum, dan standar deviasi.

```python
df_clean.describe().round(2)
```

# 8. Visualisasi Distribusi Data Numerik

Histogram digunakan untuk melihat sebaran data pada kolom numerik seperti umur, pendapatan, dan jumlah pembelian.

```python
numeric_cols = ["Age", "Income", "Purchase_Count"]

for col in numeric_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(df_clean[col], kde=True, bins=8)
    plt.title(f"Distribusi {col}")
    plt.xlabel(col)
    plt.ylabel("Frekuensi")
    plt.show()
```

# 9. Visualisasi Data Kategorik

Countplot digunakan untuk melihat jumlah data pada setiap kategori, seperti kategori produk dan status churn.

```python
categorical_cols = ["Product_Category", "Churn"]

for col in categorical_cols:
    plt.figure(figsize=(7, 4))
    sns.countplot(data=df_clean, x=col)
    plt.title(f"Jumlah Data Berdasarkan {col}")
    plt.xlabel(col)
    plt.ylabel("Jumlah")
    plt.show()
```

# 10. Encoding Data Kategorik

Data kategorik perlu diubah menjadi bentuk numerik agar dapat digunakan dalam proses Machine Learning. Pada tahap ini digunakan Label Encoding untuk kolom target dan One-Hot Encoding untuk fitur kategori produk.

```python
label_encoder = LabelEncoder()

df_encoded = df_clean.copy()
df_encoded["Churn_Label"] = label_encoder.fit_transform(df_encoded["Churn"])

df_encoded = pd.get_dummies(
    df_encoded,
    columns=["Product_Category"],
    drop_first=False
)

df_encoded
```

# 11. Memisahkan Fitur dan Target

Fitur adalah variabel yang digunakan sebagai input, sedangkan target adalah variabel yang akan diprediksi.

```python
X = df_encoded.drop(columns=["CustomerID", "Churn", "Churn_Label"])
y = df_encoded["Churn_Label"]

print("Fitur:")
display(X)

print("\nTarget:")
display(y)
```

# 12. Normalisasi dan Standarisasi Data

Normalisasi dan standarisasi digunakan untuk menyamakan skala data numerik. Hal ini penting karena beberapa algoritma Machine Learning sensitif terhadap perbedaan skala.

```python
scaler_standard = StandardScaler()
scaler_minmax = MinMaxScaler()

X_standard = X.copy()
X_minmax = X.copy()

X_standard[numeric_cols] = scaler_standard.fit_transform(X_standard[numeric_cols])
X_minmax[numeric_cols] = scaler_minmax.fit_transform(X_minmax[numeric_cols])

print("Hasil StandardScaler:")
display(X_standard.round(3))

print("\nHasil MinMaxScaler:")
display(X_minmax.round(3))
```

# 13. Membagi Dataset Menjadi Data Latih dan Data Uji

Dataset dibagi menjadi data latih dan data uji. Data latih digunakan untuk melatih model, sedangkan data uji digunakan untuk menguji performa model.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_standard,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print("Jumlah data latih:", X_train.shape[0])
print("Jumlah data uji:", X_test.shape[0])

print("\nX_train:")
display(X_train)

print("\ny_train:")
display(y_train)
```

# 14. Feature Engineering Sederhana

Feature engineering dilakukan dengan membuat fitur baru berdasarkan fitur yang sudah ada. Pada contoh ini dibuat fitur `Income_per_Purchase` untuk melihat rata-rata pendapatan terhadap jumlah pembelian.

```python
df_feature = df_clean.copy()

df_feature["Income_per_Purchase"] = df_feature["Income"] / df_feature["Purchase_Count"]

df_feature[["CustomerID", "Income", "Purchase_Count", "Income_per_Purchase"]]
```

# 15. Visualisasi Fitur Baru

Visualisasi ini digunakan untuk melihat perbandingan fitur baru `Income_per_Purchase` berdasarkan status churn.

```python
plt.figure(figsize=(8, 4))
sns.barplot(
    data=df_feature,
    x="Churn",
    y="Income_per_Purchase",
    estimator=np.mean
)
plt.title("Rata-rata Income per Purchase Berdasarkan Churn")
plt.xlabel("Churn")
plt.ylabel("Income per Purchase")
plt.show()
```

# 16. Interpretasi Hasil

Berdasarkan proses preprocessing yang dilakukan, dataset awal berhasil dibersihkan dari missing value menggunakan median. Data kategorik berhasil diubah menjadi data numerik melalui proses encoding. Data numerik juga berhasil diskalakan menggunakan StandardScaler dan MinMaxScaler.

Tahap preprocessing ini sangat penting karena kualitas data akan memengaruhi hasil analisis dan performa model Machine Learning. Dengan data yang sudah bersih, terstruktur, dan memiliki skala yang sesuai, proses pemodelan dapat dilakukan dengan lebih baik.

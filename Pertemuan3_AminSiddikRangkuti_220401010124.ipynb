```python
df = df_dirty.copy()

print("Dataset yang akan dibersihkan berasal dari:", dirty_file)
print("Shape awal:", df.shape)

display(df.head())
```

```python
print("INFO DATASET")
df.info()

print("\nSTATISTIK DESKRIPTIF")
display(df.describe(include="all"))

print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nJUMLAH DUPLIKAT")
print(df.duplicated().sum())
```

```python
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)

print("Persentase Missing Values:")
print(missing_pct[missing_pct > 0])
```

```python
print("Shape sebelum hapus duplikat:", df.shape)
print("Jumlah duplikat sebelum:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Shape setelah hapus duplikat:", df.shape)
print("Jumlah duplikat setelah:", df.duplicated().sum())
```

```python
df["kota"] = df["kota"].astype(str).str.strip().str.title()
df["kondisi"] = df["kondisi"].astype(str).str.strip().str.lower()

print("Nilai unik kolom kota:")
print(df["kota"].unique())

print("\nNilai unik kolom kondisi:")
print(df["kondisi"].unique())
```

```python
df = df.replace(["", " ", "nan", "None", "NaN"], np.nan)

print("Missing values setelah normalisasi missing:")
print(df.isnull().sum())
```

```python
df["luas_m2"] = df["luas_m2"].fillna(df["luas_m2"].median())
df["harga_juta"] = df["harga_juta"].fillna(df["harga_juta"].median())
df["kamar"] = df["kamar"].fillna(df["kamar"].mode()[0])

if df["kota"].isnull().sum() > 0:
    df["kota"] = df["kota"].fillna(df["kota"].mode()[0])

if df["kondisi"].isnull().sum() > 0:
    df["kondisi"] = df["kondisi"].fillna(df["kondisi"].mode()[0])

print("Missing values setelah imputasi:")
print(df.isnull().sum())
```

```python
def tangani_outlier_iqr(dataframe, kolom):
    Q1 = dataframe[kolom].quantile(0.25)
    Q3 = dataframe[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    outlier = dataframe[
        (dataframe[kolom] < batas_bawah) |
        (dataframe[kolom] > batas_atas)
    ]

    print(f"Kolom: {kolom}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Batas bawah: {batas_bawah}")
    print(f"Batas atas: {batas_atas}")
    print(f"Jumlah outlier: {len(outlier)}")
    print("-" * 40)

    dataframe[kolom] = dataframe[kolom].clip(
        lower=batas_bawah,
        upper=batas_atas
    )

    return dataframe
```

```python
kolom_outlier = ["harga_juta", "luas_m2", "tahun_bangun"]

for kolom in kolom_outlier:
    if kolom in df.columns:
        df = tangani_outlier_iqr(df, kolom)

print("Outlier selesai ditangani.")
```

```python
df["kamar"] = df["kamar"].round(0).astype(int)
df["tahun_bangun"] = df["tahun_bangun"].round(0).astype(int)

display(df.head())
```

```python
print("Shape akhir:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Total duplikat:", df.duplicated().sum())

assert df.isnull().sum().sum() == 0, "Masih ada missing values!"
assert df.duplicated().sum() == 0, "Masih ada data duplikat!"

print("VALIDASI BERHASIL: dataset sudah bersih.")
```
